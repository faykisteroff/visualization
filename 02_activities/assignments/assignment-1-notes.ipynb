```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)

# define the data we will be plotting
```

```python
# plotting with a legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = 'Person 1')       # data for person 1
ax.plot(x,y2, label = 'Person 2')       # data for person 2
ax.legend(loc = 'lower right')          # location of legend
```

```python
# editing the legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = "lower right",      # customizing legend now
        frameon = True,             
        fontsize = 12,              # legend font size
        ncol = 2,                   # 2 columns for legend text
        shadow = True)              # adding shadow
```

```python
# upper left corner of the legend is anchored to point 1,1 
fig, ax = plt.subplots(figsize = (5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = "upper left",
          bbox_to_anchor = (1,1))
```

```python
# plotting in a scatter 
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.text(10,95, "this value is important")       # adding a text overlay to graph
```

```python
# text annotations using transforms

fig, ax = plt.subplots()
ax.axis([0,10,0,10])        # setting the range of our x and y axis

ax.text(1,5,                          # location on figure
        ".Data: (1,5)",               # text we want to add
        transform = ax.transData      # indicates that annotation refers to data
)
ax.text(0.5,0.1, ".Axes: (0.5,0,1)",  # referring to proportion of axis dimensions
         transform = ax.transAxes)    # here refers to axes
ax.text(0.2,0.2, ".Figure: (0.2,0.2)", transform = fig.transFigure) # here refers to figure
```

```python
# annotating with arrows
# use ax.annotate function

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='lower right')

ax.annotate('this is important',    # text we are adding to graph
            xy = (10,95),           # the coord to point at      
            xytext = (20,94),        # coord of text
            arrowprops = dict(facecolor = 'pink')   # editing look of arrow to make it pink
            )      
```

```python
# editing look of arrows

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='lower right')

ax.annotate('this is important',    
            xy = (10,95),                
            xytext = (20,94),        
            arrowprops = dict(facecolor = 'red',    # editing look of arrow to make it red
                arrowstyle = 'fancy'),              # changing shape
            )      
```

```python
# axis labels

# locator to remove ticks and labels
# formatter for only labels

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator())       # removed tick marks and labels
ax.xaxis.set_major_formatter(plt.NullFormatter())   # removed labels but ticks remain
```

```python
# max limits to tick marks

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MaxNLocator(3))      # only 3 tick marks on x axis
```

```python
# set ticks to appear at specific interval

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MultipleLocator(5))     
# set ticks to appear by intervals of 5
```

```python
# rotating the appearance of tick labels

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='lower right')

plt.xticks(rotation = 45,           # seting rotation of 45 degrees for x axis ticks
           ha = 'right',            # right horizontal axis
           font = 'serif')          # serif font
```

```python
# axis title appearance changes
font1 = {'family' : 'serif', 'color' : 'deeppink'}

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='lower right')

plt.xlabel('shiny new axis',        # text for x axis
           fontsize = 17,           # font size
           fontdict = font1)        # font style
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0,       # values to range from 0-75
                       75, 
                       50)      # 50 of these values
x2 = np.array(['luffy', 'zoro', 'nami', 'usopp', 'sanji'])
y2 = np.array([110,             # values associated with our categorical data
               180, 
               240, 
               99, 
               220])

# setting the data to work with
```

```python
plt.style.use('default')

fig,(ax1, ax2) = plt.subplots(ncols = 2,        # num of columns
                              nrows = 1,        # num of rows
                              figsize = (7,3))  # figure size

ax1.scatter(x1,y1)
ax2.bar(x2, y2)
```

```python
fig,(ax1, ax2) = plt.subplots(ncols = 2,        # num of columns
                              nrows = 1,        # num of rows
                              figsize = (7,3))  # figure size

ax1.scatter(x1,y1,
    marker = "*", 
    color = 'deeppink')
ax2.bar(x2, y2,
    color = 'indigo')

ax1.set_title('pretty stars')
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],    #setting locations of axes by order
                                    ['ax2', 'ax3']],   # ax1 top left, ax2 bottom left, ax3 full right
                                    figsize = (7,4))

# specify what we want each axes object to look like

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
```

```python
# modifying figure layout

# layouts

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],   
                                    ['ax2', 'ax3']],   
                                    figsize = (7,4))

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

# setting x axis labels for each fig
someaxes['ax1'].set_xlabel('a big label', fontsize = 18)
someaxes['ax2'].set_xlabel('another label', fontsize = 18)
someaxes['ax3'].set_xlabel('label 2: 2 fast 2 furious', fontsize = 18)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],   
                                    ['ax2', 'ax3']],   
                                    figsize = (7,4),
                                    layout = 'constrained')
                                    # adding layout setting to everything fits

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel('a big label', fontsize = 10)
someaxes['ax2'].set_xlabel('another label', fontsize = 10)
someaxes['ax3'].set_xlabel('label 2: 2 fast 2 furious', fontsize = 13)
```

```python
# plots overlayed on one another

# call two plot elements on the same axes object

x = np.array(['luffy', 'zoro', 'nami', 'usopp', 'sanji'])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170,100,90,120,50])

fig, ax = plt.subplots(figsize =(7,3))

# pulling everything to same axes obj
ax.bar(x,y1,color = 'lavender')
ax.plot(x,y2, color = 'indigo')
```

```python
# adding error info - standard deviation of y2

y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))

ax.plot(x,y2,color = 'indigo')
```

```python
# use error bar function

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2,color = 'darkblue')

ax.errorbar(x,              # our x values
            y2,             # our y values 
            yerr = y2_sd,   # working with vertical (y) error bars
            fmt = 'none',           # don't plot data, just error
            color = 'darkgreen')   
```

```python
# change appearance of error bars

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2,
        color = 'violet',
        linewidth = 3)

ax.errorbar(x,              
            y2,            
            yerr = y2_sd,  
            fmt = 'none',         
            ecolor = 'darkgreen',       # error line colour
            elinewidth= 3,              # line width
            capsize= 6,                 # cap size
            capthick= 4,                # cap thickness
            errorevery = 2              # specify only every 2 bars
        )
```
