```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y)
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x,y)
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title("total growth over time")
ax.set_ylabel('total growth')
ax.set_xlabel('years since start')
```

```python
font1 = {'family':'sans-serif', 'color':'purple', 'size':20}
font2 = {'family':'monospace', 'color':'pink', 'size':14}

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title('total growth over time', fontdict= font1, loc= 'left')
ax.set_ylabel('total growth', fontdict= font2)
ax.set_xlabel('year since start', fontdict= font2)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y, marker= "*", color = 'purple')
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
    marker = "*", 
    color = 'indigo',
    linestyle = '--',
    linewidth = 1
    )
```

```python
fig, ax = plt.subplots(figsize=(7,4))
ax.plot(x,y, 
    marker = "*", 
    markersize = 7,
    color = 'deeppink',
    linestyle = '--',
    linewidth = 1,
    markeredgecolor = "#342852",
    markerfacecolor = "#de95f5"
    )

ax.grid(axis = 'y', color = 'lavender', linestyle = '--', linewidth = 4)
```
